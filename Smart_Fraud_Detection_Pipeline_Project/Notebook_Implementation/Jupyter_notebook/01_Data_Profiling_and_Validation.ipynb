### **Step 3: Load datasets**

```python
accounts = pd.read_csv("accounts.csv")
transactions = pd.read_csv("transactions.csv")
fraud = pd.read_csv("fraud_watchlist.csv")

print("Accounts:", accounts.shape)
print("Transactions:", transactions.shape)
print("Fraud Watchlist:", fraud.shape)
```

### **Step 4: Inspect datasets**

### **Step 5: Data types**

```python
print("Accounts")
print(accounts.dtypes)

print("\nTransactions")
print(transactions.dtypes)

print("\nFraud Watchlist")
print(fraud.dtypes)
```

### **Step 6: Missing values**

```python

print("Accounts Nulls")
print(accounts.isnull().sum())

print("\nTransactions Nulls")
print(transactions.isnull().sum())

print("\nFraud Nulls")
print(fraud.isnull().sum())
```

### **Step 7: Duplicate records**

```python
print("Duplicate accounts:", accounts.duplicated().sum())
print("Duplicate transactions:", transactions.duplicated().sum())
print("Duplicate fraud records:", fraud.duplicated().sum())
```

### **Step 8: Account ID duplicates**

```python
print(
    "Duplicate account IDs:",
    accounts["account_id"].duplicated().sum()
)
```

### **Step 9: Orphan accounts**

```python
account_ids = set(accounts["account_id"].dropna().astype(str).str.strip())

txn_accounts = set(
    transactions["account_id"].dropna().astype(str).str.strip()
)

orphan_transactions = txn_accounts - account_ids

print("Orphan transaction accounts:", orphan_transactions)
```

# **Conclusion**

The raw datasets were successfully profiled and validated.
The analysis identified null values, duplicate records, invalid
values, inconsistent formatting, and orphan account IDs. These
